## **Bloque 1: Gestión de Índices e Indicadores**
Trabajaremos con la lista `ipc_mensual = [0.5, 0.8, 1.2, 0.4, 0.3, 0.9, 1.1]` que representa la inflación mensual.

```python
ipc_mensual = [0.5, 0.8, 1.2, 0.4, 0.3, 0.9, 1.1]
```

```python
# 1. Agrega el dato de inflación del nuevo mes: 0.6 al final.
### TU CÓDIGO AQUÍ ###
ipc_mensual.append(0.6)
```

```python
# 2. Inserta el valor 0.7 en la segunda posición (índice 1) por un ajuste retroactivo.
### TU CÓDIGO AQUÍ ###
ipc_mensual.insert(1,0.7)
```

```python
# 3. El valor 1.2 fue un error de digitación. Cámbialo por 1.0.
### TU CÓDIGO AQUÍ ###
ipc_mensual[ipc_mensual.index(1.2)] = 1.0
```

```python
# 4. Extrae los primeros 3 meses en una lista llamada 'primer_trimestre'.
### TU CÓDIGO AQUÍ ###
primer_trimestre = ipc_mensual[:3]
primer_trimestre
```

```python
# 5. Extrae los últimos 2 meses en una lista llamada 'cierre_año'.
### TU CÓDIGO AQUÍ ###
cierre_año = ipc_mensual[-2:]
cierre_año
```

```python
# 6. Ordena la lista de menor a mayor inflación.
### TU CÓDIGO AQUÍ ###
ipc_mensual.sort()
```

```python
# 7. Encuentra el índice donde quedó el valor máximo (1.1) y guárdalo en 'pos_max'.
### TU CÓDIGO AQUÍ ###
pos_max = ipc_mensual.index(1.1)
pos_max = max(ipc_mensual)
```

```python
# 8. Elimina el valor más bajo de la lista (0.3).
### TU CÓDIGO AQUÍ ###
ipc_mensual.remove(min(ipc_mensual))
```

```python
# 9. Cuenta cuántas veces aparece el valor 0.6 y guárdalo en 'conteo_06'.
### TU CÓDIGO AQUÍ ###
conteo_06 = ipc_mensual.count(0.6)
```

```python
# 10. Invierte el orden actual de la lista.
### TU CÓDIGO AQUÍ ###
ipc_mensual.reverse()
```

```python
# --- VERIFICACIÓN AUTOMÁTICA - NO MODIFICAR ESTA CELDA ---
assert len(ipc_mensual) == 8
assert 0.3 not in ipc_mensual
assert primer_trimestre == [0.5, 0.7, 0.8]
assert cierre_año == [1.1, 0.6]
print("¡Bloque 1 completado!")
```

## **Bloque 2: Listas de Listas - Microdatos de Empresas**
Gestionarás la siguiente matriz `empresas = [['E1', 500, True], ['E2', 150, False], ['E3', 1200, True]]`.
Cada sublista corresponde a `[Nombre_Empresa, Número_Empleados, Cotiza_Bolsa]`.

```python
empresas = [['E1', 500, True], ['E2', 150, False], ['E3', 1200, True]]
```

```python
# 11. Accede al número de empleados de 'E2' y guárdalo en 'emp_e2'.
### TU CÓDIGO AQUÍ ###
posE2 = 0
for i in range(len(empresas)):
  if 'E2' in empresas[i]:
     posE2 = i

emp_e2 = empresas[posE2][1]
emp_e2
```

```python
# 12. Cambia el estado de 'E2' a True (ahora cotiza en bolsa).
### TU CÓDIGO AQUÍ ###
empresas[posE2][2] = True
```

```python
# 13. Agrega una nueva empresa: ['E4', 800, False].
### TU CÓDIGO AQUÍ ###
empresas.append(['E4', 800, False])
empresas
```

```python
# 14. Calcula el total de empresas en la matriz y guárdalo en 'n_empresas'.
### TU CÓDIGO AQUÍ ###
n_empresas = len(empresas)
n_empresas
```

```python
# 15. Crea una lista vacía 'nombres' y usa un 'for' para llenarla con los nombres (E1, E2...).
### TU CÓDIGO AQUÍ ###
nombres = []
for i in range(len(empresas)):
  nombres.append(empresas[i][0])
nombres
```

```python
# 16. Calcula el promedio de empleados y guárdalo en 'promedio_emp'.
### TU CÓDIGO AQUÍ ###
totalempleados = 0

for i in range(len(empresas)):
  totalempleados += empresas[i][1]

promedio_emp = totalempleados / len(empresas)
promedio_emp
```

```python
# 17. Elimina la sublista de la empresa 'E1'.
### TU CÓDIGO AQUÍ ###
posE1 = 0
for i in range(len(empresas)):
  if 'E1' in empresas[i]:
     posE1 = i
empresas.pop(posE1)
empresas
```

```python
# 18. Accede al nombre de la última empresa de la lista y guárdalo en 'ultima_emp'.
### TU CÓDIGO AQUÍ ###
ultima_emp = empresas[len(empresas)-1][0]
ultima_emp
```

```python
# 19. Crea una lista 'grandes' con las empresas que tienen > 1000 empleados (trae la sublista con la información completa de la o las empresas).
### TU CÓDIGO AQUÍ ###
grandes = []

for i in range(len(empresas)):
  if empresas[i][1] > 1000:
     grandes.append(empresas[i])

grandes
```

```python
# 20. Modifica el nombre de 'E2' a 'Empresa Lider' dentro de la matriz.
### TU CÓDIGO AQUÍ ###
posE2N = 0

for i in range(len(empresas)):
  if 'E2' in empresas[i]:
    posE2N = i
empresas[posE2N][0] = 'Empresa Lider'
```

```python
# --- VERIFICACIÓN AUTOMÁTICA - NO MODIFICAR ESTA CELDA ---
assert emp_e2 == 150
assert empresas[0][0] == 'Empresa Lider'
assert n_empresas == 4
assert nombres == ['E1', 'E2', 'E3', 'E4']
assert promedio_emp == 662.5
assert len(grandes) == 1
print("¡Taller finalizado exitosamente!")
```
