```python
#Área del triángulo con los lados
def areaTrian(s1:float,s2:float,s3:float)->float:

  s = (s1+s2+s3)/2

  return round((s*(s-s1)*(s-s2)*(s-s3))**(1/2),2)

areaTrian(2,3,4)
```
